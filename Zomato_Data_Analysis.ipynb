Data transformation



Remove unnecessary columns

```python
df.drop(columns=['url', 'address', 'phone'], inplace=True)
```

Rename column names

```python
df.rename(columns={
    'approx_cost(for two people)': 'cost_for_two',
    'listed_in(type)': 'restaurant_type',
    'listed_in(city)': 'city'
}, inplace=True)
```

Convert rating into numerical format

```python
df['rate'] = df['rate'].astype('string')

df['rate'] = df['rate'].str.replace('/5', '', regex=False)

df['rate'] = df['rate'].replace(
    ['NEW', '-', ''], np.nan
)

df['rate'] = pd.to_numeric(
    df['rate'], errors='coerce'
)
```

convert cost into numerical format

```python
df['cost_for_two'] = df['cost_for_two'].astype('string')

df['cost_for_two'] = df['cost_for_two'].str.replace(',', '', regex=False)

df['cost_for_two'] = pd.to_numeric(
    df['cost_for_two'], errors='coerce'
)
```

Null Value Handling

Identify missing values

```python
df.isnull().sum()
```

Find the percentage of missing values

```python
(df.isnull().sum() / len(df)) * 100
```

```python
df['rate'] = pd.to_numeric(df['rate'], errors='coerce')

df['cost_for_two'] = pd.to_numeric(
    df['cost_for_two'], errors='coerce'
)
```

```python
print(df['rate'].dtype)
print(df['cost_for_two'].dtype)

print(df[['rate', 'cost_for_two']].mean())
```

Mean

```python
# Convert cost column to float
df['cost_for_two'] = df['cost_for_two'].astype('float64')

# Perform mean imputation
df['rate'] = df['rate'].fillna(df['rate'].mean())

df['cost_for_two'] = df['cost_for_two'].fillna(
    df['cost_for_two'].mean()
)
```

verify

```python
print(df[['rate', 'cost_for_two']].dtypes)

print(df[['rate', 'cost_for_two']].isnull().sum())
```

Handle categorical null values

```python
df['location'] = df['location'].fillna(df['location'].mode()[0])

df['rest_type'] = df['rest_type'].fillna(df['rest_type'].mode()[0])

df['cuisines'] = df['cuisines'].fillna(df['cuisines'].mode()[0])

df['dish_liked'] = df['dish_liked'].fillna('Unknown')
```

verify null values

```python
df.isnull().sum()
```

Create a new column

```python
df['price_category'] = pd.cut(
    df['cost_for_two'],
    bins=[-np.inf, 500, 1000, 2000, np.inf],
    labels=['Budget', 'Moderate', 'Expensive', 'Premium']
)
```

Data Visualization



Top 10 restaurant locations (Bar Chart)

```python
df['location'].value_counts().head(10).plot(
    kind='bar', figsize=(10, 5)
)

plt.title('Top 10 Restaurant Locations')
plt.xlabel('Location')
plt.ylabel('Number of Restaurants')
plt.xticks(rotation=45)
plt.show()
```

Restaurant rating distribution (Histogram)

```python
plt.figure(figsize=(8, 5))

plt.hist(df['rate'], bins=10)

plt.title('Distribution of Restaurant Ratings')
plt.xlabel('Rating')
plt.ylabel('Number of Restaurants')
plt.show()
```

Price category distribution (Pie Chart)

```python
df['price_category'].value_counts().plot(
    kind='pie', autopct='%1.1f%%', figsize=(7, 7)
)

plt.title('Restaurant Price Category Distribution')
plt.ylabel('')
plt.show()
```

Online ordering comparison (Bar Chart)

```python
df.groupby('online_order')['rate'].mean().plot(
    kind='bar', figsize=(7, 5)
)

plt.title('Average Rating Based on Online Ordering')
plt.xlabel('Online Ordering')
plt.ylabel('Average Rating')
plt.xticks(rotation=0)
plt.show()
```

Cost vs Rating (Scatter Plot)

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['cost_for_two'], df['rate'], alpha=0.3)

plt.title('Restaurant Cost vs Rating')
plt.xlabel('Cost for Two')
plt.ylabel('Rating')
plt.show()
```
